Student Performance Tracker and Academic Risk Predictor

```python
name = input("Enter student name: ")
attendance = float(input("Enter attendance percentage: "))
maths = float(input("Enter Maths marks out of 100: "))
science = float(input("Enter Science marks out of 100: "))
english = float(input("Enter English marks out of 100: "))
print("Student details saved successfully!")
```

```python
total = maths + science + english
average = total / 3

if average >= 90:
    grade = "A+"
elif average >= 75:
    grade = "A"
elif average >= 60:
    grade = "B"
elif average >= 50:
    grade = "C"
elif average >= 40:
    grade = "D"
else:
    grade = "Fail"

print("\n--- Performance Result ---")
print("Student Name:", name)
print("Total Marks:", total)
print("Average Marks:", round(average, 2))
print("Grade:", grade)
```

```python
if average < 40 or attendance < 50:
    risk = "High Risk"
    suggestion = "Meet your teacher and focus on studies immediately."
elif average < 60 or attendance < 75:
    risk = "Medium Risk"
    suggestion = "Improve attendance and practise weak subjects."
else:
    risk = "Low Risk"
    suggestion = "Keep up the good work."

print("\n--- Risk Prediction ---")
print("Risk Status:", risk)
print("Suggestion:", suggestion)
```

```python
def get_valid_number(message, minimum, maximum):
    while True:
        try:
            value = float(input(message))

            if minimum <= value <= maximum:
                return value
            else:
                print("Please enter a value between", minimum, "and", maximum)

        except ValueError:
            print("Invalid input. Enter numbers only.")


while True:
    name = input("Enter student name: ")

    if name.strip() != "":
        break
    else:
        print("Name cannot be empty.")

attendance = get_valid_number("Enter attendance percentage: ", 0, 100)

maths = get_valid_number("Enter Maths marks out of 100: ", 0, 100)

science = get_valid_number("Enter Science marks out of 100: ", 0, 100)

english = get_valid_number("Enter English marks out of 100: ", 0, 100)

print("Student details saved successfully!")
```
